# DECISION TREE

• It is a supervised machine learning algorithm used for classification and regression.

• The model learns a set of decision rules in the form of IF–ELSE conditions.

• A Decision Tree is a flowchart-like structure where:

- The root node asks the first question.
- Each branch represents an answer.
- Internal nodes ask more questions.
- The leaf node gives the final prediction.

## Advantages
• Easy to understand

• Highly interpretable

• Handles non-linear data

• Works with numerical and categorical data

• Mimics human decision-making.

• No feature scaling required.

## Disadvantages
• Can overfit if allowed to grow too deep.

• Small changes in data may produce a different tree.

• Axis-parallel splits may not draw diagonal boundaries efficiently.

• Large trees can become complex.

## Decision Tree Algorithm
• ID3 --> Uses Entropy

• C4.5 --> Uses Gain Ratio

• CART --> Uses Gini Index (Supports both Classification and Regression.)

## Approach in Decision Tree Algorithm:
Step 1: Start with the entire dataset at the root node.

Step 2 : Evaluate all features to find the best split.

Step 3: Split the dataset based on the feature that gives maximum information gain (or minimum impurity).

Step 4: Repeat the process recursively for each branch.

Step 5: Stop splitting when a stopping condition is met.

## Impurity Measures in Decision Trees

### What is Impurity?
Impurity tells us how mixed the classes are.

- The objective of a Decision Tree is to reduce impurity after every split.

### Gini Index 
- Gini measures impurity.

Gini = 1 − (p² + q²)

p → probability of class 1

q → probability of class 2


  | Gini | Meaning     |
| ---- | ----------- |
| 0    | Pure node   |
| High | Impure node |

- Lower Gini means a better split.
  
### Entropy
Entropy measures uncertainty or disorder.

Entropy = − p log₂(p) − q log₂(q)

- Entropy ranges from 0 to 1.
  
| Entropy | Meaning          |
| ------- | ---------------- |
| 0       | Pure             |
| 1       | Completely mixed |

### Information Gain
Information Gain tells us how much impurity decreases after a split.

Information Gain =  Entropy(parent) − Weighted Entropy(children)

- Higher Information Gain means a better feature.

### Over Fitting in Decision Tree

#### Q. What is Overfitting?
Ans. A Decision Tree is built by continuously splitting the dataset into smaller and smaller groups.
If we allow the tree to keep growing without any restrictions, it may eventually memorize every training example instead of learning the underlying pattern.
This phenomenon is called Overfitting.

• Decision trees can easily overfit the training data as the depth of the tree increases.

• Overfitting occurs when the tree learns:

  -Patterns

  -Noise

  -Outliers

• Deep Tree → Overfitting

• Shallow Tree → Under fitting

### How to Control Overfitting

• Limit Tree Depth (max_depth)

• Minimum Samples Required to Split (min_samples_split)

• Minimum Samples per Leaf (min_samples_leaf)

### Prining

#### Q. What is Pruning?
Ans. Pruning means cutting unnecessary branches from the Decision Tree.
Instead of allowing the tree to grow indefinitely, we stop or trim it so that it learns only the important patterns.

• As a rule of thumb, keep the depth equal to or slightly greater than the number of features.

• Example: If you have a dataset with 100 features, a depth of 100 would still make a massive, unreadable tree. In those real-world cases, you'd likely use an ensemble method like a Random Forest, or stick to a much smaller depth (like 5 to 10) because a few key features usually do most of the heavy lifting!

### Pruning in Decision Trees

• Pruning reduces the size of the tree by removing unnecessary branches.

• Goal: Improve generalization.

• Two Types:

   -Pre-Pruning (Early stopping) : Limit tree growth during training.

- Pre-Pruning stops the tree before it becomes too complex.
The tree is not allowed to grow beyond certain limits.

• Examples:

• Max_depth

• Min_sample_split

• Min_samples_leaf

  -Post-Pruning

• In Post-Pruning, the Decision Tree is first allowed to grow completely.
After the full tree is built, unnecessary branches are removed.

#### Q. Why is Post-Pruning Often Better?
Ans. The algorithm first learns all possible patterns.
Only after evaluating the tree does it remove the branches that contribute little to prediction.
Because of this, Post-Pruning often achieves a better balance between accuracy and simplicity.


### Pre-Pruning vs Post-Pruning

| Pre-Pruning                   | Post-Pruning                            |
| ----------------------------- | --------------------------------------- |
| Stops tree growth early       | Grows the full tree first               |
| Prevents unnecessary branches | Removes unnecessary branches later      |
| Faster training               | Slower training                         |
| Risk of underfitting          | Usually better generalization           |
| Simpler implementation        | Slightly more computationally expensive |

### Hyperparameters in Decision Trees

#### Q. What are Hyperparameters?
Ans. Hyperparameters are settings that we choose before training the model.

They control how the Decision Tree is built, such as:

- How deep the tree can grow.
- When the tree should stop splitting.
- How many samples are needed before making a split.
- Which impurity measure should be used.

Unlike model parameters (which are learned from data), hyperparameters are set by the user.

• Hyperparameter tuning helps find the best combination of model parameters.

• GridSearchCV tries multiple parameter values and selects the best one using cross-validation.

### Important Hyperparameters



#### min_samples_split

• With every split number of samples also get reduce for analysis.

• Thus for higher depth problem of overfitting may occur.

• It controls the minimum number of data points a node must have before it can be split further.

• If the number of samples in a node is less than this value, the node becomes a leaf and is not split further.

###### Smaller values: The tree can grow deeper, as nodes with fewer samples can still be split. May lead to overfitting, as the model captures noise in the data.

###### Larger values: The tree grows shallower because it requires more samples to split. Can help prevent overfitting but might result in underfitting if the value is too high.



| min_samples_split                  | min_samples_leaf                        |
| ---------------------------------- | --------------------------------------- |
| Controls **when a node can split** | Controls **minimum size of every leaf** |
| Checked **before** splitting       | Checked **after** splitting             |
| Applies to the parent node         | Applies to each child node              |


### Rule for Decision tree modeling

• Rule 1 : Decision trees do not require feature scaling.

• Rule 2: Categorical variables can be encoded using Label Encoding

• Rule 3 : Check the train test accuracy as there are chance of overfitting or under fitting.

• Rule 4 : Treat imbalance data set.

• Rule 5: Use feature importance method for model simplification.

• Rule 6: Use hyper parameter tuning for deciding max_depth, min_samples_split, or min_samples_leaf

• Rule 7 : Outlier treatment because if depth is high may cause issue.

Ans. An outlier is a data point that is very different from the rest of the observations in a dataset.
It lies unusually far away from the majority of the data.

### Visualization of Boundaries
- The line (or rectangle) separating the features is called the Decision Boundary.

```python
x = df[['Culmen Length (mm)', 'Culmen Depth (mm)']] 
y = df['Species'] 
```

```python
from sklearn.tree import DecisionTreeClassifier as dtc

model = dtc(max_depth=2)
model.fit(x,y)

dtc(max_depth=2)

import numpy as np
# NumPy is needed because we're going to create thousands of points.

x_min, x_max = x.iloc[:,0].min() - 1, x.iloc[:,0].max() + 1 
y_min, y_max = x.iloc[:,1].min() - 1, x.iloc[:,1].max() + 1
# Why + and - 1?
# To leave some empty space around the graph. (Padding - taki max and min boundary ke bhr na chla jaye)

xx, yy = np.meshgrid( np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200)) 
# 200 x and y axis pe 1D array created
# graph - 2D
# meshgrid --> 1D array to 2D array for graph plotting
# np.linespace Creates equally spaced numbers.

Z = model.predict(np.c_[xx.ravel(), yy.ravel()]) 
# np.c_ combines the two arrays column-wise. gives corresponding elements
# .ravel converts a 2D array into a 1D array. (flatten --> prediction always 1D)

Z = Z.reshape(xx.shape) 
# 1D - 2d for graph

plt.figure(figsize=(8,6))

plt.contourf(xx, yy, Z, alpha=0.3)
# This fills the graph with colors.
# alpha=0.3 makes the colors transparent so the data points remain visible.

sns.scatterplot(x=x.iloc[:,0], y=x.iloc[:,1], hue=y, palette='Set1') 
# hue controls color

plt.xlabel("Culmen Length (mm)") 
plt.ylabel("Culmen Depth (mm)") 
plt.title("Decision Tree Decision Boundary") 

plt.show() 
```
