# Unsupervised Learning
- Unsupervised Learning is a type of machine learning in which the model is trained using unlabeled data. Unlike supervised learning, there are no predefined output labels (Y). The goal is to discover hidden patterns, relationships, or structures within the data.
- Think of unsupervised learning like sorting a giant basket of unsorted Lego bricks 
without instructions. Instead of a guide telling you what to build, you look at the 
bricks yourself and group them together by matching colors, shapes, or sizes.
- Basically, the computer looks at raw data, finds the similarities on its own, and 
organizes it without any human giving it the answers.

#### Key Features
- Uses unlabeled datasets.
- Finds hidden patterns and groupings automatically.
- Does not require human-labeled training data.
- Commonly used for data exploration and pattern discovery.

### Types of Unsupervised Learning
1. Clustering
    - Groups similar data points together.
    - Examples: K-Means, Hierarchical Clustering, DBSCAN.


2. Association Rule Learning
    - Finds relationships between variables.
    - Example: Apriori Algorithm (used in market basket analysis).


3. Dimensionality Reduction
    - Reduces the number of features while preserving important information.
    - Examples: PCA (Principal Component Analysis), t-SNE.

### Applications
- Customer segmentation
- Recommendation systems
- Fraud detection
- Market basket analysis
- Anomaly detection
- Image compression

### Advantages
- No need for labeled data.
- Helps discover hidden patterns.
- Useful for exploratory data analysis.
- Can handle large amounts of data.

### Disadvantages
- Results may be difficult to interpret.
- Accuracy is harder to evaluate.
- Algorithms may produce different results depending on parameters.

Suppose a shopping website has customer purchase data but no customer categories. An unsupervised learning algorithm can automatically group customers with similar buying habits, helping the business target marketing campaigns effectively.

## Clustering
- Clustering is an unsupervised learning technique where we group similar data points together into clusters, without having predefined labels.
- The basic idea is:
    - Similar data points → same cluster
    - Different data points → different clusters

#### Example
Suppose we have customer data:

We don't tell the algorithm who belongs to which group.

    - Cluster 1 → Young customers with lower spending

### Principle of Clustering
- This is the core foundation of how clustering algorithms actually "think." 
- When you throw a bunch of raw data at a computer, it doesn't have eyes to see 
groups. Instead, it relies on math and distance to decide which points belong 
together. 
- To evaluate if the computer did a good job, we use two key metrics: Intra-cluster 
distance and Inter-cluster distance. Let's break down exactly what those terms 
mean.

#### 1. Intra-Cluster Distance (Distance Within a Cluster)
- What it means: Look at a single group that has been formed. "Intra" means inside. 
So, this is the average distance between all the data points inside that same group. 

- The Goal → LOW: You want this distance to be as small as possible. If the points are 
close together, it proves they are highly similar to one another. Think of a tight-knit 
group of friends standing in a small circle whispering—they are tightly connected.

#### 2. Inter-Cluster Distance (Distance Between Clusters) 
- What it means: "Inter" means between (like an inter-state highway between 
different states). This is the distance between two different groups—usually 
measured from the center of one group to the center of another. 

- The Goal → HIGH: You want this distance to be as large as possible. If the groups are 
far apart, it proves that the clusters are genuinely different from one another.

#### The Ultimate Goal: Good Clustering 
A perfect clustering algorithm achieves both goals simultaneously: 
    
    Intra-cluste --> MINIMIZED (Tight, compact groups)
    
    Inter-cluster --> MAXIMIZED (Far apart, distinct groups)

#### Example - A Simple Analogy: School Classrooms 
Imagine a school with three different classrooms: Design, Coding, and Business. 

- Good Clustering (Low Intra, High Inter): Inside the Coding classroom, all the 
students are sitting tightly packed at their desks working together (Low Intra
cluster distance). Meanwhile, the Coding classroom is all the way down the hall, far 
away from the Design and Business rooms (High Inter-cluster distance). There is no 
confusion about who belongs where.

- Bad Clustering: The students in the Coding room are scattered wildly all over the 
school, and the Coding room itself is sharing a wall and an open doorway with the 
Design room. The groups bleed into each other, and it's impossible to tell who is 
studying what.

### Why to apply Feature Scaling in Clustering 
- The reason we must apply Feature Scaling (like Standardization or Min-Max scaling) before clustering comes down to one core concept: clustering algorithms are obsessed with distance.

- Algorithms like K-Means, Hierarchical Clustering, and DBSCAN rely heavily on calculating the mathematical distance (like Euclidean distance) between data points to decide if they belong together. 

- If you don't scale your features, the algorithm gets "blinded" by the features with the biggest numbers.

### K_Means Clustering
- K-Means Clustering is one of the most popular unsupervised machine learning 
algorithms. Its main goal is simple: group similar data points together into distinct 
clusters based on their features, without any prior labels. 

- The "K" represents the number of clusters you want to create, and "Means" refers to 
averaging the data points to find the center (centroid) of each cluster.


#### How do you choose K?
Since you have to tell the algorithm what K should be upfront, picking the right 
number can sometimes be tricky. The most common method to find the optimal K is 
the Elbow Method, where you plot the Within-Cluster Sum of Squares (WCSS) 
against different values of K and look for the "elbow" point where the drop starts to 
slow down significantly.

#### How It Works 
The algorithm uses an iterative process to find the best groupings. Here is how it walks through the data: 

1. Choose the number of clusters (K): You decide how many groups you want to find in 
your dataset (e.g., K = 3).
2. Initialize Centroids: The algorithm randomly picks K data points to act as the initial "centers" or centroids of your clusters. These are temporary centers.
3. Assign Points to the Nearest Centroid: For every single data point in your dataset, 
the algorithm calculates its distance to all K centroids (usually using Euclidean 
distance) and assigns the point to the closest one.
4. Update the Centroids: Now that the groups are formed, the algorithm calculates the 
mathematical mean (average) of all the points inside each cluster. The centroid moves to this new average position.
5. Repeat Until Convergence: Repeat Step 3 and Step 4. The algorithm keeps reassigning points and moving centroids until the centroids stop changing position, meaning the clusters have stabilized.

##### One Point to Remember 
- K-Means does NOT know what your clusters mean.
- It only give you clusters, it doesn't automatically know that these represent which groups.

#### Limitations of K-Means
- Cannot detect non-spherical clusters 
- Sensitive to outliers 
- Requires K to be predefined
- Works best with equal size clusters.


#### Example 
- Suppose we have customers represented by two features: 
- Feature 1 → Annual Spending Score 
- Feature 2 → Purchase Frequency 
- Goal: Group customers into K = 2 clusters

- Convergence

### Hierarchical Clustering
- Hierarchical Clustering is an unsupervised learning algorithm that groups similar data points into clusters by creating a hierarchy of clusters.

- The main difference from K-Means is that Hierarchical Clustering doesn't require you to choose the number of clusters at the beginning.

- The biggest advantage here is flexibility: you can decide how many clusters you 
want after the algorithm runs by looking at the resulting tree map, known as a 
dendrogram.

#### How it Works 
there are two approaches:
##### 1. Agglomerative Clustering (Bottom-Up):
    - The most common approach. Every single data point starts as its own individual cluster. The algorithm progressively merges the closest pairs of clusters together until all points are grouped into one massive cluster. 
Individual points
       ↓
Small clusters
       ↓
Larger clusters
       ↓
One big cluster

This is what you will usually mean when someone says Hierarchical Clustering.
##### 2. Divisive Clustering (Top-Down):
    - The reverse approach. All data points start in a single giant cluster, and the algorithm recursively splits the most dissimilar elements apart until every point is its own isolated cluster.
One big cluster
       ↓
Split into smaller clusters
       ↓
Split again
       ↓
Individual points



#### What is a Dendrogram?
A dendrogram is a tree-like diagram that shows which clusters were merged and at what distance.






#### Linkage
- The rule used to measure the distance between two clusters.
- When hierarchical clustering builds clusters, it repeatedly merges two clusters
- together. 
- To decide which clusters should merge, we must define how the distance between
- clusters is calculated. 
- That rule is called linkage. 
- Linkage = the mathematical rule used to compute distance between clusters in hierarchical clustering.

###### Single Linkage:
- Measures the distance between the two closest points in different clusters. (Can
- result in long, trailing chain-like clusters). 

##### Complete Linkage: 
- Measures the distance between the two farthest points in different clusters. (Produces more compact, spherical clusters). 

##### Average Linkage: 
- Calculates the average distance between all pairs of points. 

##### Ward's Linkage: 
- Minimizes the total within-cluster variance. It seeks to keep cluster sizes tightly packed and is often the default choice in implementation.

#### What makes Ward Special?


- Ward linkage merges clusters that cause the smallest increase in total within-cluster 
variance. 
- Ward produces: 
    - Compact clusters 
    - Spherical clusters 
- Works best with Euclidean distance

## DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
- It is an unsupervised clustering algorithm that forms clusters based on the density of data points and can identify outliers/noise.
-  It groups tightly packed points together and naturally flags isolated points in low-density regions as outliers (noise).

##### The key idea is:
    Dense areas become clusters, while isolated points are treated as noise.

#### Dense vs Sparse Regions
- Dense Region: Many points close together → cluster
- Sparse region: Few points → noise / outliers



- Example shown: C1 → dense cluster, C2 → dense cluster, Sparse region → noise 
- This is why DBSCAN detects outliers automatically.

### Hyper Parameters of DBSCAN 
To understand DBSCAN, you only need to know two tuning knobs: 
- Epsilon (eps): This defines the maximum distance within which points are considered neighbors.
  
- Minimum Points (minPts): The minimum number of points required within the eps 
radius to form a dense region. 

- MinPts = 4, eps = 1 

- Meaning: Inside radius 1 → at least 4 points required.


#### Dense Region
- A hypersphere or circle of radius eps that contains atleast minpts is named as dense region.


#### Types of Points i DBSCAN
1. Core Point (The Leaders)
- What it is: A point that has at least minimum points within its eps neighborhood.
- The Rule: If you draw a circle around this point with a radius of eps, it must contain 
at least MinPts (Minimum Points) inside that circle. 
- Role: These points are the building blocks and foundations of your clusters.

2. The Border Point (The Followers)
- What it is: A point that is on the fringe or edge of a crowd. 
- The Rulle: It that doesn't have enough neighbors to be a core point itself, but isclose enough to a core point to belong to its cluster.
- Role: It gets included in the cluster because it's close to a leader, but it isn't
dense enough to expand the cluster any further on its own.

3. Noise Point (The Outsiders) 
- What it is: An isolated data point standing all by itself in a quiet area. 
- The Rule: It is neither a core point nor a border point. It doesn't have enough neighbors to start a group, and it isn't close enough to any existing core point to be dragged into theirs. 
- Role: It does not belong to any cluster at all. The algorithm flags these as outliers
and ignores them.

#### Example

- Distance between S and T , R and Q = 1 
- Distance between T and U = 0.5 
- MinPts= 4 
- Eps = 1

1. The Core Points (Q, P, S) To be a Core Point, the circle around the point must contain ≥4 points total. 
- Point P: If you look at the circle centered at P, it is packed with points (several 
purple 'x' marks plus P itself). It easily meets the ≥4 threshold. Thus, P is Core. 
- Points Q and S: Similarly, the circles around Q and S capture enough neighboring 
points within their radius of 1 to meet or exceed the MinPts requirement of 4. Thus, Q and S are Core.

2. The Border Points (R, T) To be a Border Point, a point must have fewer than 4 points in its circle, but it must be within the radius of an existing Core Point. 
- Point R: Look at the topmost circle centered at R. It only contains 2 points (the yellow dot R and one purple 'x'). Because 2<4, it cannot be a Core point. However, R is close enough to fall inside the circle of Core Point Q (the distance between R and Q is 1). Thus, R is Border.

3. The Noise Points (U, V, M) To be a Noise Point, a point must fail both rules: it doesn't have 4 points in its circle, and it doesn't touch any Core Point. 
- Point U: The circle centered at U only has 2 points inside it (2<4, so not Core). While it is close to T (distance is 0.5), T is not a Core point—it's just a border point. Since U does not touch any Core point, U is Noise.
- Point V: The circle around V contains 3 points total (3<4). It sits completely isolated from the main vertical chain of Core points on the left. Since it doesn't touch a Core point, V is Noise. 
- Point M: The circle around M contains 3 points total (3<4). Just like V, it is
completely isolated from any dense cluster or Core points. Thus, M is Noise.

### DBSCAN Algorithm 
- Step 1: Label every point as: core, border, noise 
- Step 2: Remove noise points. These belong to sparse regions. 
- Step 3: For each core point:
  
    1. Create a new cluster 
    2. Add all points density-connected to it
    3. Clusters grow by density connectivity.
- Points connected through density form clusters: 
- P1 → P2 → P3 → P4 → P5 
- P5 → 6


#### Advantages of DBSCAN 
- Detects arbitrary shaped clusters 
- Handles noise automatically 
- No need to specify number of clusters (K)

#### Limitations
- Hard to choose eps. 
- Not good when cluster densities vary 
- Struggles with high dimensional data

## Silhouette Score
- Silhouette Score is a metric used to evaluate how well the data points have been clustered.
- In simple words, it tells us whether a data point is closer to its own cluster than to other clusters.
- It is especially useful with K-Means and other clustering algorithms.
- In unsupervised machine learning, you don't have true labels to check your work against, so the Silhouette Score acts as your quality check.
- It measures two things simultaneously:  
    - Cohesion (Intra-cluster similarity): Points in the same cluster should be close. 
    - Separation (Inter-cluster distance): Points in different clusters should be far.
#### Range
The Silhouette Score ranges from: -1 to +1

| Score           | Meaning                                              |
| --------------- | ---------------------------------------------------- |
| **Close to +1** | Clusters are well separated ✅                        |
| **Around 0**    | Clusters overlap or points are on boundaries         |
| **Close to -1**    | Points may have been assigned to the wrong cluster ❌ |

- So generally, Higher Sillhoutte Score = Better Defined Score

### The Math Behind the Score
- For every individual data point i in the dataset, the algorithm calculates two 
distances:

1. Intra-cluster distance: The average distance between point i and all other points belonging to the same cluster. You want this number to be as small as possible (high cohesion).
2. Nearest-cluster distance: The average distance between point i and all points in the nearest neighboring cluster. You want this number to be as large as possible (high separation).

## Association Rule Mining Algorithm 
- Association Rule Mining is an unsupervised learning technique used to discover relationships or patterns between items in a dataset.
- It aims to discover interesting patterns, correlations, or associations among a set of items or events. 
- Means if a person buy X item, what is the probability that he will also buy Y item. This is association rule mining.

##### An association rule is typically written as an If/Then statement: 
                    Antecedent(X) → Consequent(Y)

### The 3 Core Metrics You Must Know 
- To separate meaningful patterns from random coincidences, the algorithm uses 
three essential mathematical metrics:

1. Support
- Measures how frequently a specific combination of items appears in the entire dataset.

- High Support means the itemset is genuinely popular across your whole dataset.

2. Confidence
- Measures how often the rule turns out to be true. It calculates the conditional probability that a customer buys item Y, given that they have already picked up item X. 

- High Confidence means if a customer buys X, they are highly likely to also buy Y .

3. Lift
- Measures the strength of the rule compared to if the two items were purchased 
completely at random.

#### Main Algorithms
1. Apriori
- Apriori finds frequent itemsets and then generates association rules from them.
- The algorithm can tell us how frequently this combination occurs and how strong the relationship is.

2. FP-Growth
- FP-Growth (Frequent Pattern Growth) also finds frequent itemsets, but it uses an FP-tree instead of generating huge numbers of candidate itemsets like Apriori.
- Generally, it is more efficient than Apriori for large datasets.



- Lift>1: Items X and Y are positively correlated. Buying X heavily increases the chances of buying Y. 
- Lift=1: The items are completely independent. Buying X has zero impact on buying Y. 
- Lift<1: The items are negatively correlated. Customers who buy X are actively less likely to buy Y.

##### So in Association Rule Mining:
    - Support = frequency
    - Confidence = reliability
    - Lift = strength of association


### Association Rule Example 
- Rule: Soda → Orange Juice 
- Meaning: If Soda is purchased, OJ is likely purchased 



#### Goal of Association Rule Mining 
- Find rules with: 
- minimum support 
- minimum confidence

#### Applications: 
- Market Basket Analysis (MBA) is the business application of Association Rule 
Mining used to discover purchasing patterns in transactional data. 
- Cross-Selling: Retailers can use association rules to recommend complementary 
products (e.g., if a customer buys bread, they might take butter too.
- Targeted Marketing: Companies can identify groups of customers also be interested 
in butter). who frequently purchase certain items together, allowing for tailored 
marketing campaigns.

#### Evaluation Metrics 
##### Precision 
- Precision = Relevant Recommended Items / Total Recommended Items 
- Meaning: Out of the items recommended by the model, how many are actually 
relevant to the user? 

##### Recall 
- Recall : Proportion of relevant items that are successfully recommended. 
- Recall = Relevant Recommended Items / Total Relevant Items. 
- Means: Out of all items user likes, how many did we find. 

##### MAP (Mean Average Precision) 
- Precision at multiple positions in ranking. 
- MAP rewards correct ordering of recommendations 

##### NDCG (Normalized Discounted Cumulative Gain) 
- Not all positions are equally important. 
- Top recommendations matter more. 
- Higher weight for top-ranked items 
- Lower weight for lower-ranked items 
- Correct recommendations at top = more valuable



#### Limitations
- Association rule recommendations have problems. 
- Cold Start Problem: New users have no data. 
- Data Sparsity: Few transactions → difficult to find patterns. 
- Scalability: Large datasets require faster algorithms.

## Principal Component Analysis(PCA) 
- PCA is an unsupervised learning technique used for dimensionality reduction.
- In simple words: PCA reduces the number of features in a dataset while trying to preserve as much important information (variance) as possible.
- PCA finds new directions in data where variance is maximum and projects the data onto those directions to reduce dimensions.

### What is Principal Component?
- A Principal Component (PC) is a new feature created by combining the original features.
- for eg. PC1 = 0.5X1 + 0.3X2 + 0.7X3 + ...
- The coefficients tell us how strongly each original feature contributes to that component.
- The Important Order:
    - PC1 → captures the maximum variance
    - PC2 → captures the maximum remaining variance while being uncorrelated with PC1
    - PC3 → captures the next maximum remaining variance
    - And so on.

### What does 'Variance' mean here?
- In PCA, variance is basically used as a measure of how much information/variation is present in the data.

### Basic Steps of PCA
Original Dataset

       ↓

Standardize the data

       ↓

       ↓

Find eigenvalues & eigenvectors

       ↓

       ↓

       ↓

Reduced Dataset

### Why Use PCA?
- Eliminates Multicollinearity: It removes highly correlated features by merging them 
into new, independent features. 
- Speeds Up Algorithms: Fewer features mean machine learning models train and run 
significantly faster. 
- Data Visualization: It simplifies high-dimensional data (e.g., 50 features) down to 2D 
or 3D so you can plot it and visually see clusters or patterns.

### Curse of Dimensionality 
- Higher dimension take lot of time to make prediction. 
- Visualization becomes difficult 
- Goal: Reduce dimensions while keeping maximum information

#### Geometric Intuition of PCA 


- Price depends on multiple factors 
- If you want to analyze how house traits affect value, you are forced to look at multiple columns simultaneously. If you try to plot this, you need a 3D graph (one
axis for Rooms, one for Groceries, one for Price). As you add more features (like square footage, crime rate, age of the house), the dimensions explode, making the data impossible to visualize. 
- The Problem (Redundant & Correlated Dimensions): If you map rooms to the x-axis and grocery to the y-axis, you will realize that your information is spread across multiple correlated dimensions. For example, larger suburban houses might naturally have more rooms and be closer to large grocery complexes. Because these features move together, treating them as completely separate, rigid axes creates redundant information.
- The Goal of PCA: Instead of sticking with the standard, strict x and y grid lines, PCA's goal is to find new axes where the data variance (the spread of the data points) is maximum. It dynamically tilts and shifts the grid lines until it finds the line that cuts through the longest, widest stretch of your data cloud. 
- The Result (PC1): That specific direction of absolute maximum spread becomes Principal Component 1 (PC1).

#### Variance Intuition (Understanding the Signal)
- Example datasets: Consider three datasets 
- Dataset A: -1, 0, 1 
- Dataset B: -5, 0, 5 
- Dataset C: -10, 0, 10 
- The Shared Trait: All three datasets have the exact same average (Mean = 0). If you only look at the average, you cannot tell these groups apart. 
- The Difference: The Variance (how spread out the points are from zero) scales up. Dataset A has small variance, B has medium, and C has large variance.
- The Key Idea: In data science, higher variance = more information (signal). If data points are tightly bunched together, it's hard to distinguish patterns. When data points are widely spread out, it is much easier to separate, classify, and extract meaningful signals. Therefore, PCA actively tries to maximize this variance during data compression.

#### Rotate Axes to Capture Maximum Variance 
- This section transitions the concept from numbers into the visual geometry we just talked about: 
- Original Axes: The dataset starts plotted on standard feature coordinates, like Rooms and grocery. 
- New Axes: Instead of discarding one of these real axes, PCA creates entirely new, synthetic axes called PC1 and PC2.
- The Ultimate Target: The algorithm rotates the grid until the first principal component (PC1) points directly along the specific diagonal path where the data points have the absolute maximum variance (widest spread). 
- The Result: By aligning PC1 with this maximum spread, it successfully captures the most information from the original features in a single dimension.

### One Shot Summary


#### Why we need PCA 2 when we already have PCA 1 of max variance?
- PC1 captures the most information, but it rarely captures all of it. If you drop everything after PC1, you are throwing away the remaining pieces of the puzzle.
- Imagine your data is spread like an ellipse. 
- PC1 (horizontal line) captures the largest variation. 
- PC2 (vertical line) captures the remaining variation that PC1 misses.


##### Example 1 When PC1 is enough 
- Suppose: 
- PC1 explains 98% of the variance. 
- PC2 explains 2%. 
- Then you can safely keep only PC1.

##### Example 2: When PC2 is needed 
- Suppose: 
- PC1 explains 70% 
- PC2 explains 25% 
- PC3 explains 5% 
- If you keep only PC1: 
- You lose 30% of the information. 
- If you keep PC1 + PC2: 
- You retain 95% of the information. 
- This is why PCA usually keeps multiple principal components, not just one.

### Tips for Tuning eps and min_samples:

```python
from sklearn.neighbors import NearestNeighbors
# This helps us find the nearest neighboring points for every data point.
import numpy as np
import matplotlib.pyplot as plt

# Assume scaled_features is your normalized data
neighbors = NearestNeighbors(n_neighbors=5)
# For every data point, find its 5 nearest neighbors
neighbors_fit = neighbors.fit(scaled_features)

# distances is 2D: (num_samples, n_neighbors)
distances, indices = neighbors_fit.kneighbors(scaled_features)
# This calculates the nearest neighbors for every data point.
# It gives us two things:
    # distances: The actual distances to the nearest neighbors.
    # indices: The indexes of those neighboring points.

# You only need distances for this graph.

# We want the distance to the 5th neighbor (index 4)
k_distances = np.sort(distances[:, 4])

# Plot
plt.figure(figsize=(8, 5))
plt.plot(k_distances)
plt.title("K-distance Graph for DBSCAN")
plt.xlabel("Points sorted by distance")
plt.ylabel("Distance to 5th Nearest Neighbor")
plt.grid(True)
plt.show()
```

### Optimization of Epsilon vlaue

```python
from sklearn.metrics import silhouette_score
import numpy as np

best_eps = None
best_score = -1

print("eps\tSilhouette Score")

for eps in np.arange(0.2, 1.0, 0.05):
    dbscan = DBSCAN(eps=eps, min_samples=5)
    labels = dbscan.fit_predict(scaled_features)

    # Remove noise points for this particular eps
    mask = labels != -1
    X_core = scaled_features[mask]
    labels_core = labels[mask]

    if len(set(labels_core)) > 1:
        score = silhouette_score(X_core, labels_core)

        print(f"{eps:.2f}\t{score:.3f}")

        if score > best_score:
            best_score = score
            best_eps = eps

    else:
        print(f"{eps:.2f}\tNot enough clusters")

print(f"\nBest eps: {best_eps:.2f} with Silhouette Score: {best_score:.3f}")
```

- Apriori is the machine learning/data-mining algorithm used for Association Rule Mining.

        - Which items frequently occur together, and what rules can be formed from them?
        - (e.g., "People who buy bread and butter are 80% likely to also buy milk").

```python
from apyori import apriori
```

```python
rules = apriori(transaction, 
                min_support=0.015,    #Keep itemsets that appear in at least 1.5% of transactions. (How often does this combination happen?)
                min_confidence=0.3,   #Keep rules with at least 30% confidence.
                min_lift=2,           #Keep only rules with Lift ≥ 2 (A lift of 2 means the items occur together twice as often as would be expected based on the consequent's overall frequency.)
                min_length=2)         #Only consider itemsets containing at least 2 items.
```

### why FP-Growth is faster than Apriori.
- Apriori problem:
    - Generate candidates
    - Scan database multiple times
    - High computational cost
    - Try every possible combination of items
    - If store has 100 products:possible itemsets ≈ 2^100
    - Which is enormous

- FP-Growth idea:
    - Compress transactions into FP-tree
    - Extract frequent patterns directly
    - size of FP-tree, but roughly:𝑂(𝑁×𝑀)
    - Where: N = transactions, M = items per transaction
    - So it is much faster in practice.

 
- Apriori → candidate generation

- FP-Growth → pattern growth
  
| Feature               | Apriori   | FP-Growth |
| --------------------- | --------- | --------- |
| Candidate generation  | Yes       | No        |
| Database scans        | Many      | Only 2    |
| Worst-case complexity | (O(N2^M)) | (O(NM))   |
| Speed                 | Slow      | Fast      |



### Apriori suffers from exponential candidate generation, 
### while FP-Growth reduces the problem using an FP-tree and usually runs close to linear # time with respect to the dataset size.
