#### SPLIT THE DATA INTO TRAIN AND TEST

```python
from sklearn.model_selection import train_test_split as tts

x_train, x_test, y_train, y_test = tts(x, y, test_size = 0.2, stratify = y, random_state = 500)
# stratify ensures that the class distribution in the training and testing sets is similar to the original dataset.
# stratify=y is mainly used for classification tasks. It is generally not used for regression because the target values are continuous rather than belonging to discrete classes.
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

test = scaler.transform(test)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)

print(model.score(x_train, y_train))

# as R2 score is not close to 1, the data in non-linear
```

## SUPPORT VECTOR MACHINE

```python
from sklearn.svm import SVC

# kernel - linear
svm_linear = SVC(kernel = 'linear')

svm_linear.fit(x_train, y_train)
```

```python
y_pred_train_linear = svm_linear.predict(x_train)
y_pred_test_linear = svm_linear.predict(x_test)
```

| Model         | Solves | Kernel Support   | When to Use                  |
| ------------- | ------ | ---------------- | ---------------------------- |
| `SVC()`       | Dual   | Yes              | Small/medium data, nonlinear |
| `LinearSVC()` | Primal | No (linear only) | Large datasets               |

```python
print(y_train.shape)
```

```python
# kernel - sigmoid

svm_sigmoid = SVC(kernel = 'sigmoid')
svm_sigmoid.fit(x_train, y_train)
y_pred_train_sigmoid = svm_sigmoid.predict(x_train)
y_pred_test_sigmoid = svm_sigmoid.predict(x_test)
```

```python
#kernel - poly
svm_poly = SVC(kernel='poly')
svm_poly.fit(x_train, y_train)
y_pred_train_poly = svm_poly.predict(x_train)
y_pred_test_poly = svm_poly.predict(x_test)
```

```python
#kernel - rbf
svm_rbf = SVC(kernel='rbf')
svm_rbf.fit(x_train, y_train)
y_pred_train_rbf = svm_rbf.predict(x_train)
y_pred_test_rbf = svm_rbf.predict(x_test)
```

# GRID SEARCH CV

```python
from sklearn.model_selection import GridSearchCV as gscv
```

```python
SVC()
```

```python
param_grid = {'C':[0.1,1,2,10,100], 'gamma':[1,0.1,0.01,0.001]}

grid = gscv(SVC(),param_grid=param_grid,cv = 5, refit=True)
grid.fit(x_train, y_train)
grid_predict = grid.predict(x_test)
print(accuracy_score(y_test, grid_predict))

# Grid Search to find the best hyperparameters for an SVM (Support Vector Machine) model.

# C controls how much the model tries to classify every training point correctly

# Small C
# C = 0.1
# Wider margin
# Allows some mistakes
# Simpler model
# Less overfitting

# Large C
# C = 100
# Tries to classify every point correctly
# Smaller margin
# More complex model
# May overfit

# Gamma controls how far the influence of one training point extends.

# Small Gamma
# gamma = 0.001
# A point influences a large area.
# Decision boundary is smoother

# Large Gamma
# gamma = 1
# Each point influences only a small neighborhood.
# Decision boundary becomes more complex.
```
