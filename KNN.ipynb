```python
# Data Preprocessing
# split the data into ind and dep variable

x = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
# HeatMap to check MULTICOLINEARITY
# we can use VIF also

plt.figure(figsize=(15,15))
sns.heatmap(df.corr(), annot = True, cmap = 'coolwarm')
plt.show()
```

```python
# SPLITING DATA INTO TEST AND TRAIN

from sklearn.model_selection import train_test_split as tts
x_train, x_test, y_train, y_test = tts(x, y, test_size=0.25, random_state=101)
# without random_state, every time you run the code, diifferent rows will go to train and test
```

## BUILDING KNN MODEL

```python
from sklearn.neighbors import KNeighborsClassifier as knc
```

## Choose best K-value (Hyperparameter Tunning)

```python
error_rate = []
for i in range(1,40):
    knn = knc(n_neighbors = i)
    knn.fit(x_train, y_train)
    pred_i = knn.predict(x_test)
    error_rate.append(np.mean(pred_i != y_test))
```

```python
error_rate
```

```python
plt.figure(figsize=(16,16))
plt.plot(range(1,40), error_rate, color = 'blue', linestyle = 'dashed', marker = 'o', markerfacecolor = 'red', markersize = 10)
plt.title("Error Rate vs K_hyperparam_value")
plt.xlabel("K_HyperParam_Value")
plt.ylabel("Error Rate")
plt.show()
```

```python
# Basis analysis, error_rate would be less when we choose k=27

knn = knc(n_neighbors = 27)
knn.fit (x_train, y_train)
```

```python
y_pred_train = knn.predict(x_train)
y_pred_test = knn.predict(x_test)
```

```python
# EVALUATE THE MODEL

from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
```

```python
print(confusion_matrix(y_train, y_pred_train))
print("**********")
print(confusion_matrix(y_test, y_pred_test))
```

```python
print(classification_report(y_train, y_pred_train))
print("*********"*7)
print(classification_report(y_test, y_pred_test))
```

```python
print("Training Accuracy :", accuracy_score(y_train, y_pred_train))
print("**********"*5)
print("Test Accuracy :", accuracy_score(y_test, y_pred_test))
```

## CROSS VALIDATION

```python
from sklearn.model_selection import cross_val_score as cvs

training_accuracy = cvs(knn, x_train, y_train, cv = 10)
test_accuracy = cvs(knn, x_test, y_test, cv = 10)

print("Training Accuracy after CV :", training_accuracy.mean())
print("**********"*5)
print("Test Accuracy after CV :", test_accuracy.mean())
```

```python
# Why use Cross Validation?
# Imagine you split the data only once.
# Maybe you got lucky and the test data was easy, giving 96% accuracy.
# Cross-validation repeats training and testing multiple times on different splits, giving a more reliable estimate of how the model is likely to perform on unseen data.
```
