## IMPUTATIONS OF MISSING VALUE

```python
# Generally in a Numeric continuous column we fill the missing values with mean or median
# In categorical column we fill with mode
```

```python
data.nunique()
```

```python
num_col=[]
cat_col=[]

for i in data.columns:
    if data[i].nunique()>5 and data[i].dtype==float or data[i].dtype==int:
        num_col.append(i)
    else:
        cat_col.append(i)
```

```python
num_col
```

```python
cat_col
```

```python
data.dtypes
```

```python
def null_impute_numeric(col):
    data[col].fillna(data[col].mean(), inplace=True)
```

```python
def null_impute_cat(col):
    data[col].fillna(data[col].mode()[0], inplace=True)
# [0] is thr first value which is higghest as mode is in order --> first valur of every colummn
```

```python
for i in num_col:
    null_impute_numeric(i)
```

```python
data.isna().sum()
```

```python
for i in cat_col:
    null_impute_cat(i)
```

## TREATMENT OF OUTLIERS

```python
num_col.remove('Inquiries_in_the_Last_6_Months')
num_col.remove('Employment_Length')

# there is no diversity in these, categorical features
```

```python
num_col
```

```python
q3 = data.Amount_Requested.quantile(.75)
q3
```

```python
q1 = data.Amount_Requested.quantile(.25)
q1
```

```python
iqr = q3 - q1
iqr
```

```python
ul = q3 + 1.5*iqr
ul
```

```python
ll = q1 - 1.5*iqr
ll
```

```python
data.Amount_Requested.clip(ll, ul, inplace = True)
# clip is the alternative of np.where
```

```python
data.Amount_Requested.describe()
```

## OUTLIER CAPPING

```python
def out_cap(col):
    q3 = data[col].quantile(.75)
    q1 = data[col].quantile(.25)
    iqr = q3 - q1
    ll = q1 - 1.5*iqr
    ul = q3 + 1.5*iqr

    data[col] = data[col].clip(ll, ul)
```

```python
for col in num_col:
    out_cap(col)
```
