# TIME SERIES FORECASTING

- Time series data is data collected sequentially over time.
- For eg.
| Month | Sales |
| ----- | ----: |
| Jan   |   100 |
| Feb   |   120 |
| Mar   |   130 |
- The important thing is that the observations have a time order.

### Key Idea
- Past values can help us predict future values.

#### Important Properties
- Observations are not independent
- Order of data matters
- Data must be recorded at regular intervals

### Types of Data
1. Cross-Sectional Data
- Many entities are measured at one point in time.
- Example: Customer dataset
- Many people, one time.
  
2. Time Series Data
- One entity is measured across time.
-  Example: Monthly sales of one store
- One thing, many times.
  
3. Panel Data
- Many entities measured acorss time.
- Example: Sales of multiple stores across months
- Many things, many times.

## What is Time Series Forecasting?
- Time Series Forecasting is a statistical technique used to predict future values based on previously observed values in a dataset.
- It is particularly useful in various fields, such as finance, economics, weather forecasting, inventory studies, and many other areas where data is collected over time.
- Example;
    - You have:

        Jan = 100
      
        Feb = 120
      
        Mar = 130

        Apr = ?
    - Your model tries to estimate the value of April.

### Components of Time Series
1. Trend:
- The long-term direction in the data (upward, downward, or stable)
  
2. Seasonality:
- Regular patterns that repeat at specific intervals.
- Examples : Summer sales of ice cream, Retail sales during festivals
  
3. Cyclicity:
-  Fluctuations that occur at irregular intervals, influenced by economic or other external factors.
- Example: Expansion → Recession → Recovery

4. Noise:Random variation that cannot be explained by trend, seasonality, or cyclicity.

#### Difference btw seasonality and cyclicity
- Seasonality:
    - Repeats at a specific interval.
 
-  Cyclicity:
    - Fluctuations happen, but not necessarily at a fixed interval.
 


### Additive vs Multiplicative Model

#### Additive Model
Yt = Level+Trend+Seasonality+Noise
- Used when the seasonal variation remains approximately constant
- Example:
    - Year 1 seasonal increase = +20
    - Year 2 seasonal increase = +20
    - Year 3 seasonal increase = +20
- The seasonal effect stays roughly the same size.

#### Multiplicative Model
Yt = Level×Trend×Seasonality×Noise
- Used when seasonal variation increases with the trend.
- Example:
    - Small sales → small seasonal variation
    - Large sales → large seasonal variation

- Where ***level*** : Avg value of  the series

### Model Training
- During ****training****, we ALWAYS use actual values, not predicted ones (actual historical values.)
- Because we want the model to learn from true history, not from its own mistakes.




-  During ****forecasting (future)****, we may use predicted values
-  Because when you're actually predicting the future, you don't know the future actual value yet.

### Simple Exponential Smoothing (SES)
- No trend + No seasonality
- Its equation is:
    - L_t = αY_t+(1−α)L_t−1
| Symbol    | Meaning                      |
| --------- | ---------------------------- |
| (Y_t)     | Actual value at current time |
| (L_t)     | Current smoothed value       |
| (alpha)  | Smoothing parameter (0 < α ≤ 1)|
| (L_{t-1}) | Previous smoothed value      |

- We are taking a weighted average of current value and past estimate
- α → weight to current data
- (1-α) → weight to past trend



### Holt's Model: Double Exponential Smoothning
- Have trend but no seasonality.

### Holt's-Winters Model: Triple exponential smoothing model
- Have trend and seasonality.

### Auto Regressive Model: Statistical Model
- AR = AutoRegressive
- The current value depends on previous values.
- Example: today depends on yesterday

### Moving Average Model
- MA = Moving Average model
- MA uses past errors and corrects previous mistakes.

#### AR → past values
#### MA → past errors

### Auto Regressive Moving Average Model (ARMA)
- Combination of AR + MA
- Used when:
    - Data is stationary
    - No trend
    - No seasonality

### Auto Regressive Integrated Moving Average Model (AR+I+MA)

| AR        | Past values  |
| I         | Differencing |

- Handles non-stationary data.

### ADF Test
- ***ADF test = Augmented Dickey-Fuller*** test to check stationarity.
- A series is stationary when its statistical properties, especially its mean, remain roughly constant over time.

#### What does ADF actually test?
- The ADF test has two hypotheses:
    - H₀ (Null hypothesis): The series has a unit root, meaning it is non-stationary.
    - H₁ (Alternative hypothesis): The series is stationary.


- Evaluation is based on ****p-value****

| p-value      | Conclusion                             |
| ------------ | -------------------------------------- |
| **p > 0.05** | Fail to reject H₀ → **Non-stationary** |
| **p < 0.05** | Reject H₀ → **Stationary**             |

### Autocorrelation
- Is the current value related to previous values?

- ***Autocorrelation Function (ACF)***
    - ACF looks at correlation between observations and their previous lags.

- ***Partial Autocorrelation Function (PACF)***
    - PACF looks at the relationship with a particular lag after accounting for intermediate lags.

### Step 6: Identify Model Parameters using ACF & PACF

**ACF (AutoCorrelation Function)**

Measures correlation of current value with past values

**Used to identify:**

q → Non-seasonal MA

Q → Seasonal MA

**Interpretation**

Early lags (1,2,3…) → q

Seasonal lags (s, 2s, … e.g., 12, 24) → Q

**PACF (Partial AutoCorrelation Function)**

Measures direct correlation after removing intermediate effects

**Used to identify:**

p → Non-seasonal AR

P → Seasonal AR

**Interpretation**

Early lags → p

Seasonal lags (12, 24…) → P

**Parameters to Identify**

**Trend (Non-Seasonal Part)** 

| Parameter | Meaning                          |
| --------- | -------------------------------- |
| p         | AR order (past values influence) |
| d         | Differencing (removes trend)     |
| q         | MA order (past errors influence) |


**Seasonal Part**

| Parameter | Meaning                                     |
| --------- | ------------------------------------------- |
| P         | Seasonal AR                                 |
| D         | Seasonal differencing (removes seasonality) |
| Q         | Seasonal MA                                 |
| s         | Seasonal period (e.g., 12 for monthly data) |


**What These Parameters Actually Mean**

| Parameter | Meaning                                        |
| --------- | ---------------------------------------------- |
| p         | Uses past values (short-term memory)           |
| d         | Removes trend (makes data stationary)          |
| q         | Uses past errors (short-term error correction) |
| P         | Uses seasonal past values                      |
| D         | Removes seasonal pattern                       |
| Q         | Uses seasonal past errors                      |
| s         | Length of seasonal cycle                       |

**Key Understanding**

ACF → q, Q (errors)
PACF → p, P (values)


**Lag Interpretation**

| Lag Type  | Parameter |
| --------- | --------- |
| Lag 1,2,3 | p, q      |
| Lag 12,24 | P, Q      |

**Important Notes**

ACF/PACF should be applied on stationary data

d removes trend, D removes seasonality

p,q → short-term behavior

P,Q → seasonal behavior


- Lag = how far back we're looking.
- ACF = overall correlation
- PACF = direct correlation after accounting for intermediate lags

#### ACF Plot :  
- ACF measures how the current value is related to its past values (lags).
 
- ACF(k) = 𝐶𝑜𝑟𝑟(𝑌𝑡,𝑌𝑡−𝑘) , where k = lag
 
- **ACF answers:** How much does today depend on yesterday, last week, last month…?
  
 
| Lag    | Meaning             |
| ------ | ------------------- |
| Lag 1  | Today vs Yesterday  |
| Lag 2  | Today vs 2 days ago |
| Lag 12 | Monthly seasonality |

- **Role of ACF**
- Helps identify q (MA order)
- Detects seasonality
- Shows overall autocorrelation pattern


#### PACF Plot: 
- PACF measures correlation between current value and a lag after removing intermediate effects.
- PACF(k) = 𝐶𝑜𝑟𝑟(𝑌𝑡,𝑌𝑡−𝑘∣𝑌𝑡−1,𝑌𝑡−2,...,𝑌𝑡−𝑘+1)
- PACF answers: Does lag k directly influence today.
- PACF shows direct correlation after removing indirect effects.
- Role of PACF: Helps identify p (AR order)

- **Auto Regressive(AR) Model** :  “Future depends on past values”

 

- **Moving Average(MA)Model** :  “Future depends on past errors (mistakes)”

```python
for param in trend_pdq:
    for param_seasonal in seasonal_pdq:
# So for every non-seasonal combination, Python tries every seasonal combination.
      
        try:
            model = SARIMAX(train,
                                order=param,
                                seasonal_order=param_seasonal,
                                enforce_stationarity=False, 
                                enforce_invertibility=False)
# Here param and param_seasonal are not fixed numbers anymore.
# They change during every loop.

# AIC = Akaike Information Criterion.
# For model selection:
# Lower AIC is generally better.
# It balances: Good model fit + Model complexity
# So we don't simply choose the model with the most parameters.
# We're looking for a model that fits well without unnecessary complexity.
            
            result_2 = model.fit()
            print("ARIMA{}x{} - AIC:{}".format(param, param_seasonal, result_2.aic))
            
            if result_2.aic < best_aic:
                best_aic = result_2.aic
                best_param = param
                best_seasonal_param = param_seasonal
        except Exception as e:
            print("Error:", e)
            continue                                           
```

```python
print("\nBest Model:")
print("Order:", best_param)
print("Seasonal Order:", best_seasonal_param)
print("Best AIC:", best_aic)

# AIC ki value jitni negative utna best model
# Lower AIC = Better balance between accuracy and simplicity.
```

| Feature       | Lag Plot                | Durbin-Watson         |
| ------------- | ----------------------- | --------------------- |
| **Type**      | Visual                  | Statistical           |
| **Data used** | Original series         | Residuals             |
| **Output**    | Pattern                 | Numeric value         |
| **Detects**   | General autocorrelation | Lag-1 autocorrelation |

- Lag plot is a visual method that examines autocorrelation in the original time series, while Durbin-Watson is a statistical measure that primarily detects first-order (lag-1) autocorrelation in residuals.
